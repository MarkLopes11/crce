```python
import numpy as np
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
```

```python
# Step 2: Define documents and initial query
documents = [
    "Artificial intelligence is a branch of computer science.",
    "Machine learning is a subfield of AI.",
    "Deep learning allows computational models that are composed of multiple processing layers.",
    "Artificial intelligence and machine learning are revolutionizing industries.",
    "Cooking recipes and food blogs."
]

query = "artificial intelligence"
```

```python
vectorizer = TfidfVectorizer()
doc_vectors = vectorizer.fit_transform(documents)
query_vector = vectorizer.transform([query])
```

```python
# Step 4: Simulate user feedback with cosine similarity
cosine_similarities = cosine_similarity(query_vector, doc_vectors).flatten()
ranked_doc_indices = cosine_similarities.argsort()[::-1]

# Top 2 are relevant, last 1 is non-relevant
relevant_indices = ranked_doc_indices[:2]
non_relevant_indices = ranked_doc_indices[-1:]

# Extract document vectors
DR = doc_vectors[relevant_indices].toarray()
DN = doc_vectors[non_relevant_indices].toarray()
q0 = query_vector.toarray()[0]
```

```python
# Step 5: Define Rocchio parameters
alpha = 1.0
beta = 0.75
gamma = 0.15
```

```python
# Step 6: Rocchio Algorithm
rocchio_query = alpha * q0 + beta * np.mean(DR, axis=0) - gamma * np.mean(DN, axis=0)
```

```python
# Step 7: Ide Regular Formula
ide_regular_query = alpha * q0 + beta * np.sum(DR, axis=0) - gamma * np.sum(DN, axis=0)
```

```python
# Step 8: Ide Dec-Hi Formula
worst_dn_index = cosine_similarities[non_relevant_indices].argmin()
worst_dn = DN[worst_dn_index]
ide_dec_hi_query = alpha * q0 + beta * np.sum(DR, axis=0) - gamma * worst_dn
```

```python
# Step 9: Print query vectors
print("Original Query Vector (q0):")
print(q0)

print("\nRocchio Updated Query:")
print(rocchio_query)

print("\nIde Regular Query:")
print(ide_regular_query)

print("\nIde Dec-Hi Query:")
print(ide_dec_hi_query)
```

```python
# Optional: Re-retrieve using updated query vector
updated_query_vector = rocchio_query.reshape(1, -1)
new_similarities = cosine_similarity(updated_query_vector, doc_vectors).flatten()
new_ranking = new_similarities.argsort()[::-1]

print("New document ranking based on expanded query:", new_ranking)
```

## Conclusion

### 🔹 Rocchio Algorithm
- **Merit:** Balances original query and feedback using tunable weights.
- **Application:** General-purpose query refinement in IR systems.
- **Demerit:** May dilute term importance when averaging document vectors.

### 🔹 Ide Regular
- **Merit:** Emphasizes relevant documents more strongly (no averaging).
- **Application:** Suitable when relevant documents are highly informative.
- **Demerit:** Can overweight terms if too many relevant docs are used.

### 🔹 Ide Dec-Hi
- **Merit:** Strongly penalizes misleading non-relevant terms.
- **Application:** Best for precision-focused searches.
- **Demerit:** Too aggressive; may overcorrect based on one bad doc.

### ✅ Summary
Rocchio offers a balanced update, Ide Regular is best with strong relevant docs, and Ide Dec-Hi is ideal when minimizing false positives is critical.
