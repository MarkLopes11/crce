```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.datasets import load_diabetes
```

### Load Dataset

```python
data = load_diabetes()
df = pd.DataFrame(data.data, columns=data.feature_names)
df['target'] = data.target
```

### Understand the structure

```python
df.info()
```

```python
df.head()
```

### Handle Missing Values

```python
df.isnull().sum()
```

### Remove Duplicates

```python
df.duplicated().sum()
```

```python
df = df.drop_duplicates()
```

### Summary Statistics

```python
df.describe()
```

### Univariate Analysis

```python
### Histogram for age
plt.figure(figsize=(4,6))
sns.histplot(df['age'], kde=True, bins=20)
plt.title('Age Distribution')
plt.show()
```

```python
### Histogram for target = disease propagation
plt.figure(figsize=(4,6))
sns.histplot(df['target'], kde=True, color="green")
plt.title('Disease Progression Distribution')
plt.show()
```

### Bivariate Analysis

```python
### BMI vs Target
plt.figure(figsize=(4,6))
sns.scatterplot(x='bmi', y='target', data=df)
plt.title('BMI vs Disease Progression')
plt.show()
```

```python
### Blood Pressure vs Target
plt.figure(figsize=(4,6))
sns.scatterplot(x='bp', y='target', data=df)
plt.title('Blood Pressure vs Disease Progression')
plt.show()
```

### Multivariate Analysis

```python
### Correlation Matrix
plt.figure(figsize=(10,6))
corr = df.corr()
sns.heatmap(corr, annot=True, cmap='coolwarm', fmt='.2f')
plt.title('Correlation Matrix')
plt.show()
```

### Outlier Detection

```python
### Boxplots
plt.figure(figsize=(10,6))
sns.boxplot(df.drop('target', axis=1))
plt.title('Boxplots of Features')
plt.xticks(rotation=45)
plt.show()
```

### Analyze target variable

```python
plt.figure(figsize=(4,6))
sns.histplot(df['target'], kde=True, color="blue", bins=20)
plt.title('Disease Progression Distribution')
plt.show()
```

### **1. What is Exploratory Data Analysis (EDA)?**

**EDA (Exploratory Data Analysis)** is the process of visually and statistically analyzing a dataset to:

* Understand key characteristics
* Detect **patterns**, **anomalies**, and **outliers**
* Gain insights before building ML models
* Prepare the data for further processing

---

### **2. Why is EDA important in healthcare data analysis?**

EDA helps in:

* Identifying **trends** (e.g., disease risk by age or gender)
* Detecting **missing values** or **errors** in clinical records
* Understanding **relationships** (e.g., BMI vs Glucose levels)
* Revealing **class imbalance** in disease datasets
* Ensuring that model inputs are **valid**, **clean**, and **relevant**

---

### **3. Common Python Libraries Used for EDA in Healthcare:**

| Library          | Use Case                                           |
| ---------------- | -------------------------------------------------- |
| **pandas**       | Data manipulation and summaries                    |
| **matplotlib**   | Basic visualizations (line, bar)                   |
| **seaborn**      | Advanced statistical plots                         |
| **numpy**        | Numerical operations                               |
| **scikit-learn** | Dataset loading, correlation checks, preprocessing |

---

### **4. Challenges in Performing EDA on Healthcare Data:**

* **Missing or Incomplete Data**
* **Inconsistent Units** (e.g., mg/dL vs mmol/L for glucose)
* **Data Privacy & Anonymization** concerns
* **High Dimensionality** (many lab tests, vitals, scans)
* **Unstructured Medical Text** (doctor’s notes, discharge summaries)
